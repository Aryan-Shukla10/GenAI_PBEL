```python
file = open("/content/drive/MyDrive/Colab Notebooks/New Text Document.txt", "r")
p= "/content/drive/MyDrive/Colab Notebooks"
with open(p+"/New Text Document.txt","w+") as f4:
  f4.write ("kanika, 99 \n Rohit, 85 \n riya, 9 \n")
  f4.seek(0)
  content = f4.read()
  print (content)
  c=0
  f=0
  for line in content.splitlines() :
      name, marks = line.split(", ")
      if int (marks ) >= 45:
          print (name, "has passed the exam with marks:", marks)
          c=c+1
      else:
          f=f+1
          print (name, "has failed the exam with marks:", marks)
print("Number of students who passed: ", c)
print ("Number of students who failed: ", f)
```

```python
file = open("/content/drive/MyDrive/Colab Notebooks/salary.txt", "r")
p= "/content/drive/MyDrive/Colab Notebooks"
with open(p+"/salary.txt","w+") as s:
  s.write( "Rahul, 5000 \n Anjali, 60000 \n ansh, 55000 \n")
  s.seek(0)
  content = s.read()
  print(content)
  c=0
  for line in content.splitlines():
    name,salary = line.split(",")
    if(int(salary.strip()) >= 50000):
      print(name,"has a salary of:",salary)
      c=c+1
      print("Number of employees with salary >= 50000:", c)
```

```python
file = open("/content/drive/MyDrive/Colab Notebooks/attendance.txt.txt", "r")
p= "/content/drive/MyDrive/Colab Notebooks"
with open(p+"/attendance.txt.txt","w+") as s:
  s.write( "Rahul:A \n Anjali:A \n ansh:p \n")
  s.seek(0)
  content = s.read()
  print(content)
  a=0
  p=0
  for line in content.splitlines():
    name, at = line.split(":")
    if((at.strip()) == 'A'):
      print(name, "is absent")
      a=a+1
    else:
      p=p+1
      print(name, "is present")
  print("Number of absent employees:",a)
  print("Number of present employees:",p)
```
