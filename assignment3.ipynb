```python
import pandas as pd
import numpy as  np
df=pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Products.csv')
df
```

```python
average_price = np.mean(df['Price'])
print("average price =",average_price)
```

```python
def above_average(data,avg):
  print("Products Price above average:")
  for i in range(len(data)):
    if data["Price"][i]>avg:
      print(data["Product"][i],"-",data["Price"][i])
above_average(df,average_price)
```

```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/day,temperature.csv')
df
```

```python
print(df)
```

```python
highest = np.max(df['temperature'])
lowest = np.min(df['temperature'])
average = np.mean(df['temperature'])
print("highest temperature =",highest)
print('lowest temperature=',lowest)
print('average temperature=',average)
```

```python
def classify(temp):
  if temp < 20:
    return "cold"
  elif temp <=30:
    return "moderate"
  else:
    return "hot"

for i in range(len(df)):
  day = df['day'][i]
  temp = df['temperature'][i]
  category = classify(temp)
  print(day, "-", temp, "-", category)
```
